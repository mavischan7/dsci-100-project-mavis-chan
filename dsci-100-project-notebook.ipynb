#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
tournament_summary <- tournament |>
    select(-tourney_finish) |>
    summarize(across(reg_wins_pct:conf_wins_pct, list(mean = mean, min = min, max = max), na.rm = TRUE))
tournament_summary
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
# code to show the number of missing values for each column
tournament_missing <- tournament |>
    summarize(
        reg_wins_pct_missing = sum(is.na(reg_wins_pct)),
        conf_wins_pct_missing = sum(is.na(conf_wins_pct)),
        tourney_finish_missing = sum(is.na(tourney_finish)) 
    )
tournament_missing

# code to drop the NA values in conf_wins_pct and opening_round_loss
tournament_tidy <- tournament |>
    filter(!is.na(conf_wins_pct)) |>
    filter(tourney_finish != "opening_round_loss")
head(tournament_tidy)
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.3: Summarize over levels of a categorical variable (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates mean of one numeric variable for each of the levels of another categorical variable. 
- Show the outputted table. 
</div>

```r
tournament_mean <- tournament_tidy |>
    group_by(tourney_finish) |>
    summarize(reg_wins_pct_mean = mean(reg_wins_pct), conf_wins_pct_mean = mean(conf_wins_pct))
tournament_mean
```

```r
# use reg_wins_pct as numeric variable for this histogram
tournament_numeric_plot <- tournament_tidy |>
    ggplot(aes(x = reg_wins_pct)) +
    geom_histogram(binwidth = 3) +
    xlab("Regional Wins Percentage") +
    ylab("Count") +
    theme(text = element_text(size = 15)) +
    ggtitle("Proportion of Regional Win Percentages")
tournament_numeric_plot
```

```r
# use tourney_finish as categorical variable for this bar plot
tournament_categorical_plot <- tournament_tidy |>
    ggplot(aes(y = fct_infreq(tourney_finish))) +
    geom_bar() +
    xlab("Count") +
    ylab("Tourney Finish Result") +
    theme(text = element_text(size = 15)) +
    ggtitle("Proportion of Tourney Finish Categories")
tournament_categorical_plot
```

Visualization 1: Scatter plot that plots conference wins percentage vs regional wins percentage. This is relevant because the predictor variable is regional wins percentage and the response variable is conference wins percentage. I want to see the relationship between both quantitative variables, so that I can determine whether I should use KNN or linear regression.

```r
tournament_scatterplot <- tournament_tidy |>
    ggplot(aes(x = reg_wins_pct, y = conf_wins_pct)) +
    geom_point() +
    xlab("Regional Wins Percentage") +
    ylab("Conference Wins Percentage") +
    theme(text = element_text(size = 15)) +
    ggtitle("Conference Wins Percentage vs \nRegional Wins Percentage")
tournament_scatterplot
```

Visualization 2: Bar plot that compares average regional wins percentage and tourney finish. While this plot does not directly help to answer the research question, this plot helps to see if the percentage of regional wins help predict team quality as reflected by the tourney finish.

```r
tournament_bar <- tournament_tidy |>
    group_by(tourney_finish) |>
    summarize(reg_wins_pct_mean = mean(reg_wins_pct)) |>
    ggplot(aes(x = reg_wins_pct_mean, y = fct_reorder(tourney_finish, reg_wins_pct_mean))) +
    geom_bar(stat = "identity") +
    xlab("Average Regional Wins Percentage") +
    ylab("Tourney Finish") +
    ggtitle("Tourney Finish vs Regional Wins Percentage") +
    theme(text = element_text(size = 15))
tournament_bar
```
